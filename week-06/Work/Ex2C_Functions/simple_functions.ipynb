Working with defining custom functions

```python
def favorite_things(name, movie, music):
    print(name)
    print(movie)
    print(music)
```

```python
favorite_things("Jeremiah", "Chronicles", "Tyler the Creator")
```

```python
def why_im_here(purpose):
    print(purpose)
```

```python
why_im_here("The reason I joined Year Up United was because \nI wanted to enter a career where I know I will \nbe able to better support my wife's dreams.")
```

```python
def fav_place(place, why):
    return f"One of my favorite places is {place} because {why}"
```

```python
fav_place("Jack in the Box", "I would go there whenever my mood was down and it was a place I would go to hang out with some old friends.")
```

```python
fav_place("Great America", "I would go here all the time with family and friends.")
```
