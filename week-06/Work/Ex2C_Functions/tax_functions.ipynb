Combining multiple functions to display a tax breakdown.


```python
def soc_sec_tax(income, tax, total):
    print(f"Because you gross is ${income: .2f} \n"
          f"your Social Security Tax will withhold ${tax: .2f}\n" 
          f"making your total income ${total: .2f}")

pay = float(input("What is your monthly gross $"))
ss_tax = pay * 0.062
monthly_total = pay - ss_tax

soc_sec_tax(pay, ss_tax, monthly_total)
```

```python

def medicare_tax(pay, tax, total):
    print(f"Because you get paid {pay: .2f} \n" 
          f"Medicare will withhold {tax: .2f} \n"
          f"Which will make your total income ${total: .2f}")

income = float(input("What is your monthly income $"))
med_taxes = income * 0.0145
full_total = income - med_taxes

medicare_tax(income, med_taxes, full_total)
```

```python

annual_pay = float(input("What is your Weekly Gross: $"))
code = int(input("What is your Withholding Code: "))

def federal_tax(income, code):
    if code == 0:
        rate = 0.23
    elif code == 1:
        rate = 0.21
    elif code == 2:
        rate = 0.195
    elif code == 3:
        rate = 0.185
    else:  # code >= 4
        rate = 0.18 - 0.005 * (code - 4)

    fed_taxes = income * rate
    total = income - fed_taxes

    print(f"Your Monthly Income is ${income:.2f}\n"
          f"You've Entered Code: {code} for Tax Withholding\n"
          f"Your Tax Withholding is ${fed_taxes:.2f}\n"
          f"Your Net Total is ${total:.2f}")

# call the function
federal_tax(annual_pay, code)


        
```
