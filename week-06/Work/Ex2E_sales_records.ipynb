### Exercise 2.E Using functions for analyzing data
This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
sales_data = [
('Laptop', 'East', 1200.00),
('Phone', 'West', 850.00),
('Tablet', 'East', 400.00),
('Monitor', 'South', 650.00),
('Keyboard', 'West', 75.00),
('Laptop', 'South', 1150.00),
('Phone', 'East', 900.00),
('Tablet', 'West', 425.00),
('Monitor', 'East', 680.00),
('Keyboard', 'South', 80.00),
('Laptop', 'West', 1300.00),
('Phone', 'West', 820.00),
]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    return sum(amt for _, r, amt in data if r == region)
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    return sum(1 for _, r, amt in data if r == region) 
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'West'))
print(region_count(sales_data, 'South'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):
    total = region_total(data, region)
    count = region_count(data, region)
    return total / count if count != 0 else 0
```

```python
print(region_average(sales_data, 'East'))
print(region_average(sales_data, 'West'))
print(region_average(sales_data, 'South'))
print(region_average(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    return [record for record in data if record[2] > threshold]
```

```python
print(sales_above(sales_data, 500))
```

```python
print(sales_above(sales_data, 1))
```

## Lab 5. Top and bottom sale

```python
def top_sale(data):
    return max(data, key=lambda s: s[2])
def bottom_sale(data):
    return min(data, key=lambda s: s[2])
```

```python
print(top_sale(sales_data))
```

```python
print(bottom_sale(sales_data))
```

```python
print(f"The top selling product was a {top_sale(sales_data)[0]}\n" 
      f"in the {top_sale(sales_data)[1]} region \n"
      f"for ${top_sale(sales_data)[2]}")
```

## Lab 7. Unique regions

```python
def unique_regions(data):
    return sorted({r for _, r, _ in data})
    
```

```python
print(unique_regions(sales_data))
```

As you can see the using this function brings up unique region names and removes 
duplicates. Making it easier to see how many unique regions there are. 
If I were to use this function to compare overall cash total by region, this would be much
easier than looking at each store in a certain region.

## Lab 8. Region summary

```python

def region_summary(data, region):
    region_data = [r for r in data if r[1] == region]

    total = region_total(region_data, region)
    count = region_count(region_data, region)
    avg = region_average(region_data, region)
    top = top_sale(region_data)  # full tuple

    return {
        "Region": region,
        "Count": count,
        "Total": total,
        "Average": avg,
        "Top Sale": f"${top[2]} ({top[0]})" if top else "None"
    }
```

```python
print(region_summary(sales_data, "West"))
```

## Lab 9. Full summary

```python

def region_summary(data, region):
    region_data = [r for r in data if r[1] == region]

    total = region_total(region_data, region)
    count = region_count(region_data, region)
    avg = region_average(region_data, region)
    top = top_sale(region_data)

    return {
        "Region": region,
        "Count": count,
        "Total": total,
        "Average": avg,
        "Top Sale": f"({top[0]}) ${top[2]}"
    }

    
```

```python

for region in unique_regions(sales_data):
    summary = region_summary(sales_data, region)
    print(f"\nRegion: {summary['Region']}")
    print(f"Count: {summary['Count']}")
    print(f"Total: ${summary['Total']}")
    print(f"Average: ${summary['Average']:.2f}")
    print(f"Top Sale: {summary['Top Sale']}")
```
