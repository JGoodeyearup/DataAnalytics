## Lab 1. Working With Names - Parsing Strings

```python
test_name = "Chris Bellington"
```

```python

def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == " ":
            first = name[:i]
            last = name[i+1:]
            print(f"Full name:{name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
        
```

```python
parse_and_display(test_name)
```

```python
test_list = ["Chris Redfield", "Leon Kennedy", "Mia Redfield"]
```

```python
def parse_and_display(list):
    for name in list:
        for i in range(len(name)):
            if name[i] == " ":
                first = name[:i]
                last = name[i+1:]
                print(f"Full name: {name}")
                print(f"First name: {first}")
                print(f"Last name: {last}")
```

```python
parse_and_display(test_list)
```

```python
full_name = "Jeremiah Devon Goode"
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        first = name[:8]
        middle = name[-11:-6]
        last = name[-6:]
        print(f"Full name: {name}")
        print(f"First name: {first}")
        print(f"Middle name: {middle}")
        print(f"Last name: {last}")
```

```python
parse_and_display(full_name)
```
