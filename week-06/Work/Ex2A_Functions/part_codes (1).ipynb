# Exercise 2.A Using Bult-in Functions

## Lab 2 Parsing Strings - part codes

```python
test_product = [("nike:556-L", "addidas:212-M", "reebok:648-S")]
```

```python
def supplier_code(item):
    return item.split(":")[0]
def product_num(item):
    return item.split(":")[1].split("-")[0]
def size(item):
    return item.split("-")[1]
```

```python
for group in test_product:
    for item in group:
        print("Item:", item)
        print("Supplier code:", supplier_code(item))
        print("Product Number:", product_num(item))
        print("Product Size:", size(item))
        print()
```

```python
for group in test_product:
    for item in group:
        print(f"For {item}, the supplier is {supplier_code(item)}. \nPart number is {product_num(item)} is available in size {size(item)}.")
        
```
