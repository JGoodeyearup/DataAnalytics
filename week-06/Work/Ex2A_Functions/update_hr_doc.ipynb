This lab makes updates to information extracted from an employee database.

```python
hr_list = [
('0123', 'HR', 'Rebecca Yang', '69000'),
('0121', 'IT', 'Mark Blick', '67000'),
('0068', 'IT', 'Kahna Larsen', '112000'),
('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
for i in range(len(hr_list)):
    if hr_list[i][2] == 'Mark Blick':
        hr_list[i] = (hr_list[i][0], hr_list[i][1], 'Mark Blick-Hawley', hr_list[i][3])
```

```python
print(hr_list)
```
