```python
# Can you add a boolean to an integer? What happens?
score = 0
points = 10
answer = ("Yes", "yes", "Y", "y", "True", "true", "T", "t")
question = input("True or False: \nJay is cool??")

if question in answer:
    score += points
    print(f"You've gained {points} points!!!")
else:
    print(f"Ouch")
    print(f"Your score is: {score}")
```
