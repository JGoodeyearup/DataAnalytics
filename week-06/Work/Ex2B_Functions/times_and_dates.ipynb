Working with datetime functions

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
formatted = today.strftime("%A, %m, %d, %Y")
```

```python
print(f"Today is {formatted}")
```

# My Birthday

```python
my_bday = datetime.date(2000, 8, 23)
```

```python
birthday = my_bday.strftime("%m, %d, %Y")
```

```python
print(f"My birthday is {birthday}")
```

# The Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
day_90 = ninety_d.strftime("%m, %d, %Y")
```

```python
print(f"90 days from today will be {day_90}")
```

# Time For Dinner

```python
dinner_t = datetime.time(9, 30)
```

```python
dinner = dinner_t.strftime("%I:%M")
```

```python
print(f"Dinner time is {dinner}pm")
```
