## Series

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {
    'first': 13,
    'second': 14,
    'third': 15,
    'fourth': 16
}
```

```python
ser1 = pd.Series(aa)
print(ser1)
```

```python
ser2 = pd.Series(ab)
print(ser2)
```

```python
ser3 = pd.Series(list(ac)) #Switched to list to run properly
print(ser3) # Set types are incompatible with pd.series() b/c it's unordered
```

```python
ser4 = pd.Series(ad)
print(ser4)
```

```python
ser5 = pd.Series(17, 18, 19, 20)
print(ser5) #TypeError: Index(...) must be called with a collection of some kind, 18 was passed
```
