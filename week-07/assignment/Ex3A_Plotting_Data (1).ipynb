# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker = 'o')
plt.show()
```

```python
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x,y)
plt.show()
```

## Styling Plots

```python
plt.plot(x,y, ls = '--', c = 'r')
plt.show()
```

```python
plt.plot(x,y, marker = 's', mfc = 'g', ms = 10)
plt.show()
```

```python
plt.plot(x,y, marker = 'X', ms = 30, mfc = 'm')
plt.show()
```

```python
plt.plot(x,y, marker = '*', ms = 15, mfc = 'w')
plt.show()
```

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]

plt.title("Products")
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.bar(products,sales, color = 'skyblue', width = .5)
plt.show()
```

## Subplots

```python

# Data for sine wave
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Random data for histogram
data = np.random.randn(1000)

# Create subplots
fig, axs = plt.subplots(1, 2, figsize=(10, 4))

# First subplot: line chart
axs[0].plot(x, y)
axs[0].set_title("Sine Wave")
axs[0].set_xlabel("X values")
axs[0].set_ylabel("sin(X)")

# Second subplot: histogram
axs[1].hist(data, bins=30)
axs[1].set_title("Histogram of Random Data")
axs[1].set_xlabel("Value")
axs[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
