## Looping Through Arrays

```python
arrA = np.array([(11, 12, 13), (14, 15, 16)])
print(arrA)
```

```python
for x in arrA:
    x = x * 2
    print(x)
```

```python
for y in x:
    y = y * 2
    print(y)
```

The output in the first syntax displays the two lists doubled while the other one shows then second list doubled.

```python
for x in arrA:
    loop_results = x * 2
    for y in x:
        y = y * 2
```

```python
print(loop_results) # shows the second list doubled.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis = 1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2) #ValueError: all input arrays must have the same shape
                           
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA), axis = 1)
print(another_joined_array) #ValueError: all but the concatenation must match exactly 
```

```python
hey_stack = np.hstack((arrA, arrA)) # Displays the data along the horizon to compare
print(hey_stack)
```

```python
vi_stack = np.vstack((arrA, arrA)) # Displays data vertical 
print(vi_stack)
```

```python
d_stack = np.dstack((arrA, arrA)) # Displays everything in group both groups and separated by a space between the groups
print(d_stack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky) #Prints the list as two full separate arrays 
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino) # This print 3 arrays with the first one showing the full list 
                  # The other two arrays are two halves of the first full array
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Searching and Replacing Array Values

```python
arrB = np.array([(7, 6),(13, 5),(7, 3)])

sorted_arrB = np.sort(arrB)
print(sorted_arrB)
    
```

```python
z = np.where(arrB < 7)
arrC = arrB[z]
print(arrC)
```

```python
h = np.where(arrB > 7)
arrD = arrB[h]
print(arrD)
```
