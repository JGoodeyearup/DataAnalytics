## Filtering a DataFrame

```python
df_age = df[df['Age'] > 18]
```

```python
df_score = df[(df['Score'] > 50) & (df['Score'] < 100)]
```

```python
df_age
df_score
```

## Modifying a DataFrame

```python
df_cleaned = df.drop(df[df['Score'] < 0].index)
df_cleaned
```

```python
df_cleaned.drop(columns=["City"])
```

```python
df_cleaned.fillna(df['Age'].mean())
```

```python
df_cleaned.replace(np.nan, df['Score'].mean())
```

```python
df_cleaned.dropna()
```

```python
df_cleaned.set_index('ID')
df_cleaned.drop(["ID"], axis = 1)
df_cleaned
```
