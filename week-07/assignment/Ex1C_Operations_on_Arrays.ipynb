### How does broadcasting and using the add function actually work??

The add function combines an array with another array with the same size, displaying an outcome of both arrays added together. The size needs to be the same size so the values can stay in line with each other. If the arrays conflict in size you will receive an error message saying 
"ValueError: operands could not be broadcast together with shapes (2,1) (4,5)". I like to think of broadcast like making sure everyone has a window to the other side. If the size do not match in arrays not everyone can see.

## Teamwork
The scenario for this set of
problems is that you are volunteering with a voter registration campaign, and you
have been asked to oversee a squad of door-knockers. There are three people on
your team: Ali, Ben, and Carla. Your squad went out to knock on doors each evening
last week, Monday through Friday, and now you are assembling that data

```python
Ali = np.array([5, 4, 6, 4, 2])
Ben = np.array([0, 5, 5, 3, 4])
Carla = np.array([3, 6, 0, 3, 4])

all_knocks = np.sum([Ali, Ben, Carla])

print(all_knocks)
```

```python
per_person = np.divide(all_knocks, 3)
print(f"In the group there were {all_knocks} knocks in total \n"
      f"with {int(per_person)} knocks per person")
```

```python
all_knoc = np.add(Ali, Ben, Carla)
voters = np.divide(all_knoc, 2)
print_results = np.cumsum(voters)

print(np.ceil(print_results))
```
