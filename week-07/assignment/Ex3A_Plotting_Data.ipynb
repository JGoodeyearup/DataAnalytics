# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y, marker = 'o')
plt.show()
```

```python
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x,y)
plt.show()
```

## Styling Plots

```python
plt.plot(x,y, ls = '--', c = 'r')
plt.show()
```

```python
plt.plot(x,y, marker = 's', mfc = 'g', ms = 10)
plt.show()
```

```python
plt.plot(x,y, marker = 'X', ms = 30, mfc = 'm')
plt.show()
```

```python
plt.plot(x,y, marker = '*', ms = 15, mfc = 'w')
plt.show()
```
