# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
data = {'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East',
'West', 'North'],
'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]}

df_sales = pd.DataFrame(data)

df_sales
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
total_sold = df_sales['Units_Sold'].sum()
total_revenue = df_sales['Revenue'].sum()
print(f"{total_sold} units were sold across all products and regions, \n"
      f"generating ${total_revenue} in revenue.")
```

```python
avg_sold = df_sales['Units_Sold'].mean()
rev_avg = df_sales['Revenue'].mean()

print(f"{np.trunc(avg_sold)} units were sold generating an avg of ${format(rev_avg, '.2f')} in revenue.")
```

## Grouping Data

```python
total = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()

print(total)
```

```python
avg = df_sales.groupby('Region')[['Units_Sold', 'Revenue']].mean()
print(avg)
```

```python
region_total = df_sales.groupby(['Product', 'Region'])[['Units_Sold', 'Revenue']].sum()
print(region_total)
```

## Sorting and Ranking Data

```python
df_sales.sort_values('Units_Sold')
```

```python
df_sales.sort_values('Revenue', ascending = False)
```

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending = False)
df_sales
```

## Merging DataFrames

```python
product_data = {
    'Product': ['A', 'B', 'C'], 
    'Category': ['Electronics', 
                 'Furniture', 
                 'Appliances']
}

df_products = pd.DataFrame(product_data)
```

```python
df_merge = pd.merge(df_sales, df_products, on = 'Product')
df_merge
```

```python
df_merge.groupby('Category')['Revenue'].sum()
```
