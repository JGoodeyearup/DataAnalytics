## Generating Random Numbers

## Generating Random Choices from an Array

```python
students = np.array(['Jeremiah', 'Brandon', 'Kevin', 'Destiney', 'Edmund', 'Dahlia', 'Hritik', 'Sthefani', 'Luis', 'Natalia'])
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size= (3, 4))
print(breakout_groups) # This wouldn't work in reality because as you can see there are repeats of names within the output
```

```python
call_on_weighted = np.random.choice(students, p = [0.3, 0, 0.1, 0.1, 0.2, 0, 0.1, 0.1, 0.1, 0], size = (1, 5))
print(call_on_weighted)

# Jeremiah = 30%
# Brandon =  0%
# Kevin = 10%
# Destiney = 10%
# Edmund = 20%
# Dahlia = 0%
# Hritik = 10%
# Sthefani = 10%
# Luis = 10%
# Natalia = 0%
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc= 10, scale = 5, size =(4, 5))
print(bell_curve)
```
